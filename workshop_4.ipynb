# PARTE 1 — Arquivos CSV

## 1.1 — Criando um arquivo CSV

Vamos criar um arquivo CSV com dados fictícios de uma pequena turma de alunos.  
Usaremos o módulo `csv` nativo do Python para isso.

O arquivo terá três colunas: **nome**, **nota** e **cidade**.

````python
import csv

# Dados a serem salvos
dados = [
    ["nome", "nota", "cidade"],         # Cabeçalho
    ["Ana", 8.5, "São Paulo"],
    ["Bruno", 7.0, "Rio de Janeiro"],
    ["Carla", 9.2, "Belo Horizonte"],
    ["Diego", 6.8, "Curitiba"],
    ["Eva", 10.0, "Porto Alegre"],
]

# Abre/cria o arquivo para escrita
# 'w' = escrita; newline e encoding garantem compatibilidade
with open("alunos.csv", mode="w", newline="", encoding="utf-8") as arquivo:

    escritor = csv.writer(arquivo)

    escritor.writerows(dados)

print("Arquivo 'alunos.csv' criado com sucesso!")
print("O arquivo contém:", len(dados) - 1, "alunos (sem contar o cabeçalho).")
````

## 1.2 — Visualizando o conteúdo bruto do CSV

Antes de ler com bibliotecas especiais, vamos ver como o arquivo realmente parece por dentro.  
Usaremos o comando `open()` para ler o texto puro do arquivo.

````python
# Lê o CSV como texto puro
with open("alunos.csv", mode="r", encoding="utf-8") as arquivo:
    conteudo = arquivo.read()  # Lê todo o conteúdo de uma vez

print(conteudo)
````

## 1.3 — Leitura com o módulo `csv` nativo do Python

O Python já tem um módulo chamado `csv` que facilita a leitura de arquivos CSV.  
Ele cuida automaticamente de casos como valores que contêm vírgulas dentro de aspas.

````python
# Lê o CSV de forma estruturada com DictReader
import csv

with open("alunos.csv", mode="r", encoding="utf-8") as arquivo:

    # DictReader: cada linha vira um dicionário
    leitor = csv.DictReader(arquivo)

    for linha in leitor:
        print(f"Nome: {linha['nome']} | Nota: {linha['nota']} | Cidade: {linha['cidade']}")
````

## 1.4 — Leitura com listas (sem dicionário)

Também podemos ler o CSV linha por linha como uma lista simples.  
Nesse caso, acessamos os valores pela posição (índice) em vez do nome.

````python
# Lê o CSV como listas de valores
import csv

with open("alunos.csv", mode="r", encoding="utf-8") as arquivo:

    leitor = csv.reader(arquivo)

    # Extrai e pula o cabeçalho
    cabecalho = next(leitor)  # next() pega a próxima linha do leitor
    print("Cabeçalho:", cabecalho)
    print()

    for linha in leitor:
        # linha[0] = nome, linha[1] = nota, linha[2] = cidade
        nome  = linha[0]
        nota  = linha[1]
        cidade = linha[2]
        print(f"  {nome} tirou nota {nota} e mora em {cidade}")
````

## 1.5 — Leitura com pandas

O **pandas** é a biblioteca mais popular para trabalhar com dados em Python.  
Ela transforma o CSV em uma estrutura chamada **DataFrame** — que é como uma tabela inteligente.

# PARTE 2 — Arquivos Excel (.xlsx)

---

## O que é um arquivo Excel?

O **Excel** é um programa de planilhas da Microsoft. Os arquivos Excel terminam com **.xlsx** (versões modernas) ou **.xls** (versões antigas).

Diferente do CSV, um arquivo Excel pode ter:
- **Múltiplas abas** (também chamadas de *sheets* ou *planilhas*)
- **Formatação** (cores, fontes, bordas)
- **Fórmulas** (como `=SOMA(A1:A10)`)
- **Gráficos**

**Quando usamos Excel no mundo real?**
- Relatórios financeiros e contábeis
- Controle de estoque
- Planejamento de projetos
- Qualquer situação onde as pessoas trabalham com tabelas no dia a dia

---

## 2.1 — Criando um arquivo Excel com duas abas

Vamos criar um arquivo Excel com **duas abas** (sheets):
- **Aba 1:** `Turma_A` — com notas de uma turma
- **Aba 2:** `Turma_B` — com notas de outra turma

````python
import pandas as pd

turma_a = pd.DataFrame({
    "nome":   ["Alice", "Beto", "Cleo", "Duda"],
    "nota":   [9.0, 7.5, 8.0, 6.5],
    "cidade": ["São Paulo", "Campinas", "Santos", "Sorocaba"]
})

turma_b = pd.DataFrame({
    "nome":   ["Eduardo", "Flávia", "Gustavo", "Helena"],
    "nota":   [5.0, 9.5, 8.8, 7.0],
    "cidade": ["Rio de Janeiro", "Niterói", "Petrópolis", "Volta Redonda"]
})

# ExcelWriter: cria Excel com múltiplas abas
with pd.ExcelWriter("turmas.xlsx", engine="openpyxl") as escritor:

    # Escreve cada turma em sua aba
    turma_a.to_excel(escritor, sheet_name="Turma_A", index=False)

    turma_b.to_excel(escritor, sheet_name="Turma_B", index=False)

print("Arquivo 'turmas.xlsx' criado com sucesso!")
print("O arquivo contém duas abas: Turma_A e Turma_B")
````

## 2.2 — Listando as abas do arquivo Excel

# PARTE 3 — Arquivos JSON

---

## O que é um arquivo JSON?


É um formato muito usado para **trocar dados entre sistemas**, especialmente na internet.

O JSON é organizado de forma **hierárquica**, como uma árvore.  
Usa chaves `{}` para objetos, colchetes `[]` para listas, e dois pontos `:` para separar chave e valor.

**Exemplo de JSON:**

```json
{
  "nome": "Ana",
  "idade": 28,
  "hobbies": ["leitura", "música", "programação"],
  "endereco": {
    "cidade": "São Paulo",
    "estado": "SP"
  }
}
```

**Quando usamos JSON no mundo real?**
- APIs da internet (como o JSON que o Google Maps retorna)
- Configurações de programas e aplicativos
- Dados de redes sociais
- Comunicação entre servidores

---

## 3.1 — Dicionários Python e JSON

Em Python, usamos **dicionários** para armazenar dados em pares chave-valor.  
O JSON tem uma estrutura muito parecida — por isso a conversão é natural.

````python
# Dicionário com dados da escola
escola = {
    "nome": "Escola Python",
    "ano_fundacao": 2020,
    "ativa": True,
    "turmas": [
        {
            "turma": "A",
            "alunos": [
                {"nome": "Alice", "nota": 9.0},
                {"nome": "Beto",  "nota": 7.5}
            ]
        },
        {
            "turma": "B",
            "alunos": [
                {"nome": "Carla",   "nota": 8.5},
                {"nome": "Danilo",  "nota": 6.0}
            ]
        }
    ]
}

print("Dicionário Python criado")
print(f"Nome da escola: {escola['nome']}")
print(f"Ano de fundação: {escola['ano_fundacao']}")
print(f"Escola ativa: {escola['ativa']}")
print(f"Número de turmas: {len(escola['turmas'])}")
````

## 3.2 — Convertendo o dicionário para JSON e salvando

O módulo `json` do Python permite converter dicionários em texto JSON e salvar em arquivo.

````python
import json

# Salva o dicionário como JSON
with open("escola.json", mode="w", encoding="utf-8") as arquivo:

    # indent=2 para formatação legível; ensure_ascii=False preserva acentos
    json.dump(escola, arquivo, indent=2, ensure_ascii=False)

print("Arquivo 'escola.json' criado com sucesso!")
````

## 3.3 — Visualizando o conteúdo bruto do JSON

````python
# Lê o JSON como texto puro
with open("escola.json", mode="r", encoding="utf-8") as arquivo:
    conteudo = arquivo.read()

print("Conteúdo bruto do arquivo escola.json")
print()
print(conteudo)
````

## 3.4 — Lendo o arquivo JSON

````python
import json

# Lê o JSON e converte para dicionário Python
with open("escola.json", mode="r", encoding="utf-8") as arquivo:
    dados_escola = json.load(arquivo)  # Retorna um dicionário Python

print(f"Nome: {dados_escola['nome']}")
print(f"Ano: {dados_escola['ano_fundacao']}")
print(f"Ativa: {dados_escola['ativa']}")
print()
````

## 3.5 — Navegando pela estrutura hierárquica do JSON

O JSON pode ter estruturas aninhadas — dados dentro de dados.  
Podemos navegar por essa hierarquia acessando chaves uma por vez.

````python
import json

with open("escola.json", mode="r", encoding="utf-8") as arquivo:
    dados_escola = json.load(arquivo)

turmas = dados_escola["turmas"]

for turma in turmas:
    print(f"Turma: {turma['turma']}")

    for aluno in turma["alunos"]:
        print(f"    Aluno: {aluno['nome']} | Nota: {aluno['nota']}")
    print()
````

## 3.6 — Modificando e salvando dados JSON

````python
import json

with open("escola.json", mode="r", encoding="utf-8") as arquivo:
    dados_escola = json.load(arquivo)

# Adiciona novo aluno e salva
# índice 0 = primeira turma (Turma A)
novo_aluno = {"nome": "Fernanda", "nota": 10.0}
dados_escola["turmas"][0]["alunos"].append(novo_aluno)

with open("escola.json", mode="w", encoding="utf-8") as arquivo:
    json.dump(dados_escola, arquivo, indent=2, ensure_ascii=False)

print("Novo aluno adicionado e arquivo salvo!")
print()

print("Alunos da Turma A agora:")
for aluno in dados_escola["turmas"][0]["alunos"]:
    print(f"  {aluno['nome']}: {aluno['nota']}")
````

## 3.7 — Lendo JSON com pandas

Quando o JSON tem uma estrutura de lista (tabela plana), podemos usar o pandas para ler.

````python
import pandas as pd
import json

# JSON com estrutura de lista (ideal para DataFrame)
alunos_lista = [
    {"nome": "Alice",   "nota": 9.0, "cidade": "São Paulo"},
    {"nome": "Beto",    "nota": 7.5, "cidade": "Campinas"},
    {"nome": "Carla",   "nota": 8.5, "cidade": "Santos"},
    {"nome": "Danilo",  "nota": 6.0, "cidade": "Sorocaba"}
]

with open("alunos_lista.json", "w", encoding="utf-8") as f:
    json.dump(alunos_lista, f, indent=2, ensure_ascii=False)

df_json = pd.read_json("alunos_lista.json", encoding="utf-8")

print("JSON lido com pandas")
print()
print(df_json.to_string(index=False))
````

## 3.8 — Convertendo entre string JSON e dicionário

Além de trabalhar com arquivos, podemos converter JSON diretamente em memória.

````python
import json

# String JSON como uma API retornaria
texto_json = '{"produto": "Notebook", "preco": 2500.00, "disponivel": true}'

# json.loads(): string → dicionário
produto = json.loads(texto_json)

print("Convertendo string JSON para dicionário")
print(f"Produto: {produto['produto']}")
print(f"Preço: R$ {produto['preco']:.2f}")
print(f"Disponível: {produto['disponivel']}")
print()

# json.dumps(): dicionário → string
texto_novamente = json.dumps(produto, indent=2)
print("De volta para string JSON")
print(texto_novamente)
````

# PARTE 4 — Arquivos XML

---

## O que é um arquivo XML?

É um formato de dados baseado em **tags**, muito parecido com o HTML das páginas web.

A diferença principal: no HTML, as tags têm nomes fixos (`<p>`, `<div>`, `<h1>`).  
No XML, **você define os próprios nomes das tags** de acordo com seus dados.

**Exemplo de XML:**

```xml
<?xml version="1.0" encoding="UTF-8"?>
<escola>
  <aluno id="1">
    <nome>Ana</nome>
    <nota>8.5</nota>
    <cidade>São Paulo</cidade>
  </aluno>
  <aluno id="2">
    <nome>Bruno</nome>
    <nota>7.0</nota>
    <cidade>Campinas</cidade>
  </aluno>
</escola>
```

**Estrutura de uma tag XML:**
- `<aluno>` — tag de abertura
- `</aluno>` — tag de fechamento
- `id="1"` — atributo (informação extra dentro da tag de abertura)
- O texto entre as tags é o conteúdo (valor)

**Quando usamos XML no mundo real?**
- Nota Fiscal Eletrônica (NF-e) no Brasil
- Arquivos de configuração de muitos softwares
- Feeds RSS de sites de notícias
- Integração entre sistemas empresariais legados

---

## 4.1 — Criando um arquivo XML

Vamos criar um arquivo XML com dados de alunos usando o módulo `xml.etree.ElementTree`.  
Esse módulo já vem com o Python — não precisa instalar nada.

````python
# xml.etree.ElementTree: módulo padrão para XML
import xml.etree.ElementTree as ET

# Criação da tag principal do XML
raiz = ET.Element("escola")

alunos_dados = [
    {"id": "1", "nome": "Ana",    "nota": "8.5", "cidade": "São Paulo"},
    {"id": "2", "nome": "Bruno",  "nota": "7.0", "cidade": "Rio de Janeiro"},
    {"id": "3", "nome": "Carla",  "nota": "9.2", "cidade": "Belo Horizonte"},
    {"id": "4", "nome": "Diego",  "nota": "6.8", "cidade": "Curitiba"},
]

for aluno_info in alunos_dados:

    # Cria elemento <aluno> filho de <escola> com atributo id
    aluno_elem = ET.SubElement(raiz, "aluno", id=aluno_info["id"])

    nome_elem = ET.SubElement(aluno_elem, "nome")
    nome_elem.text = aluno_info["nome"]  # O texto vai dentro da tag

    nota_elem = ET.SubElement(aluno_elem, "nota")
    nota_elem.text = aluno_info["nota"]

    cidade_elem = ET.SubElement(aluno_elem, "cidade")
    cidade_elem.text = aluno_info["cidade"]

# Cria a árvore XML completa
arvore = ET.ElementTree(raiz)

# indent() formata com recuo (Python 3.9+)
ET.indent(arvore, space="  ")

arvore.write("alunos.xml", encoding="utf-8", xml_declaration=True)

print("Arquivo 'alunos.xml' criado com sucesso!")
````

## 4.2 — Visualizando o conteúdo bruto do XML

Assim como fizemos com o CSV e JSON, vamos abrir o arquivo como texto puro para ver sua estrutura.

````python
# Lê o XML como texto puro
with open("alunos.xml", mode="r", encoding="utf-8") as arquivo:
    conteudo = arquivo.read()

print(conteudo)
````

## 4.3 — Lendo o XML com ElementTree

Agora vamos ler o XML de forma estruturada para acessar seus dados.

````python
import xml.etree.ElementTree as ET

# Carrega e analisa o arquivo XML
arvore = ET.parse("alunos.xml")

# getroot() retorna o elemento raiz
raiz = arvore.getroot()

print("Lendo o XML com ElementTree")
print()
print(f"Tag do elemento raiz: <{raiz.tag}>")
print(f"Número de alunos encontrados: {len(raiz)}")
print()
````

## 4.4 — Percorrendo os elementos e acessando atributos

````python
import xml.etree.ElementTree as ET

arvore = ET.parse("alunos.xml")
raiz = arvore.getroot()

print("=== Dados de todos os alunos ===")
print()

for aluno in raiz:

    # .get() acessa atributos; .find().text acessa conteúdo dos filhos
    id_aluno = aluno.get("id")

    nome   = aluno.find("nome").text
    nota   = aluno.find("nota").text
    cidade = aluno.find("cidade").text

    print(f"ID {id_aluno}: {nome} — Nota: {nota} — Cidade: {cidade}")
````

## 4.5 — Buscando elementos específicos no XML

Podemos buscar elementos diretamente pelo nome da tag usando `findall()`.

````python
import xml.etree.ElementTree as ET

arvore = ET.parse("alunos.xml")
raiz = arvore.getroot()

# findall() retorna todos elementos com a tag especificada
todos_alunos = raiz.findall("aluno")

print("Buscando todos os elementos <aluno>")
print(f"Total de alunos encontrados: {len(todos_alunos)}")
print()

print("Nomes dos alunos:")
for aluno in todos_alunos:
    print(f"  - {aluno.find('nome').text}")
````

````python
import xml.etree.ElementTree as ET

arvore = ET.parse("alunos.xml")
raiz = arvore.getroot()

# Extrai notas e calcula estatísticas
notas = []

for aluno in raiz.findall("aluno"):
    # Converte string para float
    nota = float(aluno.find("nota").text)
    notas.append(nota)

media = sum(notas) / len(notas)
maior = max(notas)
menor = min(notas)

print("Estatísticas extraídas do XML")
print(f"Notas encontradas: {notas}")
print(f"Média: {media:.1f}")
print(f"Maior nota: {maior}")
print(f"Menor nota: {menor}")
````

## 4.6 — Inspecionando tags e atributos de forma genérica

Às vezes recebemos um XML desconhecido e precisamos explorá-lo antes de saber o que fazer.

````python
import xml.etree.ElementTree as ET

arvore = ET.parse("alunos.xml")
raiz = arvore.getroot()

print("Explorando a estrutura do XML")
print()
print(f"Raiz: <{raiz.tag}>")
print()

# iter() percorre todos os elementos recursivamente
for elemento in raiz.iter():

    tag_info = f"<{elemento.tag}>"

    # Adiciona atributos à representação, se houver
    if elemento.attrib:
        atributos = ", ".join(f"{k}='{v}'" for k, v in elemento.attrib.items())
        tag_info = f"<{elemento.tag} {atributos}>"

    texto = elemento.text.strip() if elemento.text and elemento.text.strip() else "(sem texto)"

    print(f"  Tag: {tag_info} | Texto: {texto}")
````
