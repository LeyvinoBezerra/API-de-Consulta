## 1. Tuplas

Propriedade principal que as diferenciam:

As tuplas -> imutáveis, por isso ocupam menos memória que as demais coleções. Isso significa que: Uma vez emitidos, os dados não podem ser alterados. É um registro fixo.

--> Utilização: use Tuplas para dados que são "configurações" ou registros que não devem mudar. Ex: coordenadas geográficas (latitude, longitude), os dias da semana, ou números de documentos

```python
# Criando uma tupla
tupla = ('Segunda', 'Terça', 'Quarta', 'Quinta', 'Sexta', 'Sábado', 'Domingo')
```

```python
# Validando seu conteúdo
tupla
```

```python
# Buscando por um elemento específico através do índice
tupla[0]
```

```python
# Buscando o índice de um elemento específico
tupla.index('Terça')
```

```python
# Percorrendo os elementos de uma tupla
for elemento in tupla:
  print(elemento)
```

```python
# Validando a imutabilidade de uma tupla
tupla.append('Feriado')
```

## 2. Listas

Propriedade principal: Podemos adicionar, remover e trocar os itens de lugar

--> Use Listas quando você tem uma coleção de itens que pode crescer ou mudar (ex: lista de mercado, lista de tarefas, placar de um jogo).

```python
# Criando listas de compras
l1 = ['Leite', 'Pão', 'Maça', 'Café', 'Ovos']
l2 = ['Chá', 'Biscoito', 'Banana', 'Suco', 'Iogurte']
```

```python
# Concatenando as listas
l3 = l1 + l2
print(l3)
```

```python
# Validando o conteúdo da lista individualmente
l2
```

```python
# Duplicando os elementos de uma lista
l2_2 = l2 * 2
print(l2_2)
```

```python
# Percorrendo os elementos de uma lista utilizando um loop
for item in l2_2:
  print(item)
```

```python
# Buscando por um elemento específico através do índice
l1[0]
```

```python
# Verificando se o índice está correto
l1
```

```python
# Buscando por um intervalo específico através do índice
l1[0:2]
```

```python
# Adicionando um elemento à lista
l1.append('Manteiga')
print(l1)
```

```python
# Removendo um elemento da lista
l1.remove('Pão')
print(l1)
```

```python
# Ordenando os elementos de uma lista
l1.sort()
print(l1)
```

```python
# Removendo o último elemento da lista (pilha)
l1.pop()
print(l1)
```

```python
# Verificando o número de elementos da lista (tamanho)
len(l1)
```

```python
# Invertendo a ordem dos elementos da lista
l1.reverse()
print(l1)
```

```python
# Deletando a lista
del(l1)
```

```python
# Checando se a lista foi deletada
print(l1)
```

## 3. Dicionários

Propriedades que os diferenciam: os dicionários conseguem armazenar por chave (key) e valor(value).

-->Use Dicionários quando precisar rotular seus dados para encontrá-los rápido (ex: informações de um usuário: {"nome": "João", "idade": 25}).

```python
# Criando um dicionário de alunos
alunos= {'João Silveira': 32, 'Carlos Ramiro': 22, 'Cristina Santos': 14}
```

```python
# Consultando o valor de um elemento específico através da chave (idade)
alunos['Cristina Santos']
```

```python
# Adicionando um novo elemento ao dicionário
alunos['Manuela Conceição'] = 11
```

```python
# Validando o conteúdo do dicionário
print(alunos)
```

```python
# Removendo um elemento do dicionário
del(alunos['João Silveira'])
print(alunos)
```

```python
# Verificando as os itens do dicionário
alunos.items()
```

```python
# Verificando as chaves do dicionário
alunos.keys()
```

```python
# verificando os valores do dicionário
alunos.values()
```

```python
# Criando um novo dicionário de alunos
alunos2 = {'Rodrigo Fernandes': 13, 'Marina Silva': 14}
print(alunos2)
```

```python
# Atualizando o dicionário de alunos com os elementos do dicionário alunos2
alunos.update(alunos2)
print(alunos)
```

```python
# Verificando a atualização do dicionário de alunos
alunos.items()
```

```python
# Percorrendo os elementos do dicionário utilizando um loop
for nomes, idades in alunos.items():
  print(f'Nome: {nomes}, idade: {idades}')
```

## 4. Bônus: Conjuntos

Propriendades principais que os diferem:

Não aceita duplicatas: se tentarmos inserir o mesmo item duas vezes, o Python simplesmente ignora a segunda tentativa.

Não tem ordem: diferente da lista, não conseguimos acessar o "primeiro" ou o "último" item via índice (ex: conjunto[0] gera erro).

--> Utilize conjuntos sempre que precisar remover itens repetidos, ou se precisar verificar o que dois grupos têm em comum.

```python
# Exemplo de conjunto (set)
biomeleculas = {'proteína', 'ácidos nucleicos', 'carboidrato', 'lipídeo',
                'ácidos nucleicos', 'carboidrato', 'carboidrato', 'carboidrato'}
```

```python
# Exibindo o conteúdo do conjunto
print(biomeleculas)
```

```python
# Validando que não é possível acessar os elementos de um conjunto por posição, pois ele é uma coleção não ordenada
try:
    print(biomeleculas[0])
except TypeError as e:
    print(f"Erro: {e} - Não podemos acessar por posição!")
```

```python
# Exemplos de conjuntos (set) e de interseção entre eles
c1 = {1,2,3,4,5}
c2 = {3,4,5,6,7}
c3 = c1.intersection(c2)
```

```python
# Elementos comuns entre os conjuntos c1 e c2
print(c3)
```

```python
# Elementos presentes no conjunto c1, mas não presentes no conjunto c2
c1.difference(c2)
```

```python
# Elementos presentes no conjunto c2, mas não presentes no conjunto c1
c2.difference(c1)
```
